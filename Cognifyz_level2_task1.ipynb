```python
import pandas as pd
df = pd.read_csv("Dataset .csv")
table_booking_percentage = (df["Has Table booking"].value_counts(normalize=True) * 100)

print(table_booking_percentage)
```

```python
online_delivery_percentage = (df["Has Online delivery"].value_counts(normalize=True) * 100)

print(online_delivery_percentage)
```

```python
average_rating_table_booking = df.groupby("Has Table booking")["Aggregate rating"].mean()

print(average_rating_table_booking)
```

```python
import matplotlib.pyplot as plt

average_rating_table_booking = df.groupby("Has Table booking")["Aggregate rating"].mean()

average_rating_table_booking.plot(kind="bar")

plt.title("Average Rating: Table Booking vs No Table Booking")
plt.xlabel("Has Table Booking")
plt.ylabel("Average Rating")
plt.show()
```

```python
online_delivery_price = pd.crosstab(
    df["Price range"],
    df["Has Online delivery"],
    normalize="index"
) * 100

print(online_delivery_price)
```

```python
import matplotlib.pyplot as plt

online_delivery_price = pd.crosstab(
    df["Price range"],
    df["Has Online delivery"],
    normalize="index"
) * 100

online_delivery_price.plot(kind="bar")

plt.title("Online Delivery Availability by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Percentage of Restaurants")
plt.legend(title="Online Delivery")
plt.show()
```
