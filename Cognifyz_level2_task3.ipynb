```python
import pandas as pd
```

```python
df = pd.read_csv("Dataset .csv")
```

```python
df["Restaurant Name Length"] = df["Restaurant Name"].str.len()
df["Address Length"] = df["Address"].str.len()

print(df[["Restaurant Name", "Restaurant Name Length",
          "Address", "Address Length"]].head())
```

```python
import matplotlib.pyplot as plt

sample = df.head(10)

plt.figure(figsize=(10,5))
plt.bar(sample["Restaurant Name"], sample["Restaurant Name Length"])
plt.xticks(rotation=90)
plt.title("Restaurant Name Length")
plt.xlabel("Restaurant Name")
plt.ylabel("Length")
plt.show()
```

```python
df["Has Table Booking"] = df["Has Table booking"].map({"Yes": 1, "No": 0})
df["Has Online Delivery"] = df["Has Online delivery"].map({"Yes": 1, "No": 0})

print(df[["Has Table booking", "Has Table Booking",
          "Has Online delivery", "Has Online Delivery"]].head())
```

```python
import matplotlib.pyplot as plt

booking_count = df["Has Table Booking"].value_counts()

booking_count.plot(kind="bar")
plt.title("Has Table Booking (Encoded)")
plt.xlabel("0 = No, 1 = Yes")
plt.ylabel("Number of Restaurants")
plt.show()
```

```python
delivery_count = df["Has Online Delivery"].value_counts()

delivery_count.plot(kind="bar")
plt.title("Has Online Delivery (Encoded)")
plt.xlabel("0 = No, 1 = Yes")
plt.ylabel("Number of Restaurants")
plt.show()
```
