```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Dataset .csv")
```

```python
price_count = df["Price range"].value_counts()

print(price_count)
```

```python
price_count.plot(kind="bar")
plt.title("Most Common Price Range")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")
plt.show()
```

```python
average_rating = df.groupby("Price range")["Aggregate rating"].mean()

print(average_rating)
```

```python
average_rating.plot(kind="bar")

plt.title("Average Rating by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Average Rating")

plt.show()
```

```python
rating_color = df.groupby("Rating color")["Aggregate rating"].mean()

print(rating_color)
```

```python
rating_color.plot(kind="bar")

plt.title("Average Rating by Rating Color")
plt.xlabel("Rating Color")
plt.ylabel("Average Rating")

plt.show()
```
